```python
# data quality

print("age >100 :",(df['Age'] > 100).sum())
print("Cart_Abandonment_Rate > 100 :", (df["Cart_Abandonment_Rate"] > 100).sum())
print("Total_Purchases < 0 :", (df['Total_Purchases'] < 0).sum())
print("Discount_Usage_Rate > 100 :", (df['Discount_Usage_Rate'] > 100).sum())
```

```python
# missing values

for col in numeric_col:
    print(f"Missing values in {col}")
    print(df.groupby(df[col].isna())['Churned'].mean())
```

```python
df.loc[df['Total_Purchases'] < 0, ['Total_Purchases','Average_Order_Value','Returns_Rate','Churned']]
```

```python
negative_purchases = df["Total_Purchases"] < 0

print("Number of invalid rows:", negative_purchases.sum())
print("Percentage of dataset:", negative_purchases.mean() * 100)

print("\nChurn rate of invalid rows:")
print(df.loc[negative_purchases, "Churned"].mean())

print("\nChurn rate of valid rows:")
print(df.loc[~negative_purchases, "Churned"].mean())
```

```python
checks = {
    "Age > 100": df["Age"] > 100,
    "Cart_Abandonment_Rate > 100": df["Cart_Abandonment_Rate"] > 100,
    "Discount_Usage_Rate > 100": df["Discount_Usage_Rate"] > 100
}

for name, condition in checks.items():
    print(f"\n--- {name} ---")
    print("Invalid rows:", condition.sum())
    print("Percentage:", condition.mean() * 100)
    print("Churn rate:", df.loc[condition, "Churned"].mean())
```
